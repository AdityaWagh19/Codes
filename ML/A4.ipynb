# A4: Gradient Descent (Local Minima of (x + 3)^2)

Dataset: `none (no dataset needed)`

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def f(x):
    return (x + 3)**2

def grad_f(x):
    return 2 * (x + 3)
```

```python
x = 2

learning_rate = 0.1
tolerance = 1e-6
max_iterations = 25

history = [x]
```

```python
for i in range(max_iterations):

    gradient = grad_f(x)

    x_new = x - learning_rate * gradient

    if abs(x_new - x) < tolerance:
        break

    x = x_new
    history.append(x)

    print(
        f"Iteration {i+1}: x = {x:.4f}, f(x) = {f(x):.4f}"
    )
```

```python
print("Local Minima x =", x)
print("Minimum Value y =", f(x))
```

```python
plt.figure(figsize=(6,4))

plt.plot(
    history,
    [f(i) for i in history],
    marker='o'
)

plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Gradient Descent Convergence")

plt.grid(True)
plt.show()
```
